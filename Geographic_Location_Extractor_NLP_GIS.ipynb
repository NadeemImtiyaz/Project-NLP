```python
import spacy
nlp = spacy.load("en_core_web_sm")
print("NLP model loaded succesfully!")
```

```python
text = """
Heavy rainfall affected India and Nepal.
Flooding was reported near the Brahamputra River.
Snowfall occured in london.
"""
print(text)
```

```python
import spacy

nlp = spacy.load("en_core_web_sm")

doc = nlp(text)

for entity in doc.ents:
  print(entity.text,"-",
        entity.label_)
```

```python
places = []

for entity in doc.ents:
  if entity.label_ in ["GPE","LOC"]:
    places.append(entity.text)

places = list(dict.fromkeys(places))

print("Extracted Locations:", places)
```
